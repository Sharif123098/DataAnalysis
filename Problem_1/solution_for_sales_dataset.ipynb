# Order Data Cleaning and Analysis

### Reading the file

```python
import pandas as pd

file_path = "pandas_dataset.csv"
df = pd.read_csv("pandas_dataset.csv")

print("Shape before cleaning:", df.shape)
df.head()
```

### Rename column+Clean the dataset+Drop rows


```python
# Remove rows that contain no data and reset the index.
df = df.dropna(how="all").reset_index(drop=True)

# Clean the column names.
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_", regex=False)
)

print("Cleaned columns:")
print(df.columns.tolist())
df.head(5)
```

### Check missing values

```python
missing_values = df.isna().sum()
display(missing_values.to_frame("missing_values"))
```

### Count and handle missing values

```python
df["customer_name"] = df["customer_name"].fillna("Unknown")

if "profit" in df.columns:
    df["profit"] = df["profit"].fillna(df["profit"].mean())

if "discount" in df.columns:
    df["discount"] = df["discount"].fillna(df["discount"].median())

print("Missing values after handling:")
display(df.isna().sum().to_frame("missing_values"))
```

### Remove duplicate orders

```python
duplicate_mask = df["order_id"].duplicated(keep=False)

print("Duplicate orders:")
display(df.loc[duplicate_mask].sort_values("order_id"))

removed_duplicates = df["order_id"].duplicated().sum()
df = df.drop_duplicates(subset="order_id", keep="first").reset_index(drop=True)

print("Duplicates removed:", removed_duplicates)
```

### Filter orders

```python
print("Orders with unit_price greater than 20,000:")
display(df[df["unit_price"] > 20000].reset_index(drop=True))
```

```python
print("Completed orders with unit_price greater than 10,000:")
display(
    df.loc[
        (df["unit_price"] > 10000) & (df["status"] == "Completed"),
        ["customer_name", "category", "status"]
    ].reset_index(drop=True)
)
```

### Create new columns

```python
df["total_amount"] = df["quantity"] * df["unit_price"]

df["customer_type"] = df["quantity"].apply(
    lambda x: "Bulk Buyer" if x >= 3 else "Regular Buyer"
)

display(
    df[["order_id", "customer_name", "quantity",
        "unit_price", "total_amount", "customer_type"]]
    .head(10)
)
```

### Final cleaned dataset with sales row removed

```python
print("Shape before cleaning:", (25, 9))
df = df.dropna(subset=["sales"])
print("Shape after cleaning:", df.shape)

print("First 10 rows:")
display(df.head(50))
```
